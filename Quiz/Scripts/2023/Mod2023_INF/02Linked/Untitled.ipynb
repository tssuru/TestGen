```python
import random
```

```python
def create(n):
    tmp=[]
    return [random.randint(0,1) for _ in range(n)]    
```

```python
create(4)
```

```python
def check(lst):
    i=0
    for el in lst:
        if el: 
            i+=1 
        else: 
            i-=1
        if i<1:
            return False
    return sum(lst)<len(lst)-2
```

```python
def fill(cnt, n):
    res=[]
    while len(res)<cnt:
        tmp=create(n)
        if check(tmp):
            tmp.append(0)
            res.append(tmp)
            
    return res        
```

```python
a = fill(20, 8)
a
```
