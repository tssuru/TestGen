```python
def load(fname, dtype=dict):
    res=[]
    with open(fname) as f:
        for line in f:
            if line.isspace():continue
            res.append(line.strip())
    it=iter(res)
    return dtype(zip(it,it))
```

```python
routes = {'Z': "``змійкою''", 'S': 'спіраллю'}
```

```python
kinds = {'n+': 'проти годинникової стрілки',
 'n-': 'за годинниковою стрілкою',
 'n+-': 'діагоналями (перпендикулярно головній діагоналі)',
 'n++': 'діагоналями (паралельно головній діагоналі)',
 'n0+': 'горизонталями',
 'n+0': 'вертикалями'}
```

```python
rk= [('S', k) for k in kinds if len(k)<3]+[('Z', k) for k in kinds if len(k)>2]
```

```python
#parts = load('what.txt')
parts = {'z00': 'центрального ромба',
 'z01': 'трикутника над головною діагоналлю',
 'z02': 'трикутника під головною діагоналлю',
 'z03': 'верхнього чверть-трикутника',
 'z04': 'правого чверть-трикутника',
 'z05': 'нижнього чверть-трикутника',
 'z06': 'лівого чверть-трикутника',
 'z07': 'трикутника над побічною діагоналлю',
 'z08': 'трикутника під побічною діагоналлю'}
```

```python
#froms = load('from.txt')
froms = {'p01': 'правого кута',
 'p02': 'лівого кута',
 'p03': 'нижнього кута',
 'p04': 'верхнього кута',
 'p05': 'нижнього лівого кута матриці',
 'p06': 'нижнього правого кута матриці',
 'p07': 'верхнього лівого кута матриці',
 'p08': 'верхнього правого кута матриці',
 'p09': 'центрального кута'}
```

```python
romb = [('z00', f'p0{i}') for i in range(1,5)]
exclude = [('z01', 'p05'),('z02', 'p08'),('z07', 'p06'),('z08', 'p07')]
tri2 = [(x, y) for x in (el for el, _ in exclude) for y in (el for _, el in exclude) if (x,y) not in exclude]
center = [(f'z0{i}', 'p09') for i in range(3,7)]
```

```python
tmp = [(f'z0{i}', f'p0{j}') for i in range(3,7) for j in range(5,9)]
tmp1 = [(x,y) for x,y in tmp 
       if ('верх' in parts[x])== ('верх' in froms[y])
       if ('нижн' in parts[x])== ('нижн' in froms[y])
      ]
tmp2 = [(x,y) for x,y in tmp 
       if ('лівог' in parts[x])== ('лівог' in froms[y])
       if ('право' in parts[x])== ('право' in froms[y])
      ]
tri4 = tmp1+tmp2

res= romb+tri2+center+tri4
```

for x,y in tri4:
    print(parts[x], 'від', froms[y])

```python
with open('parts_from.txt', 'w') as f:
    for x, y in res:
        xx = parts[x]
        yy = froms[y]
        if x=='z00': yy='його '+yy
       # elif '5'<=y[-1]<='8':
        #    yy += ' матриці'
        print(x, xx, y, yy ,'\n',sep='\n', file=f)
```

```python
with open('kind_route.txt', 'w') as f:
    for x, y in rk:
        xx = routes[x]
        yy = kinds[y]
        print(y, yy ,x, xx, '\n',sep='\n', file=f)
```
