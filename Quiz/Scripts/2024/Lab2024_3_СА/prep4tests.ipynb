```python
import re
```

```python
fname = 'lab3SA.tex'
resname = 'lab3var.txt'
i = 0
regex = '%.*'
```

```python
with open(fname) as f:
    with open(resname, 'w') as ff:
        for line in f:
            if ('begin' not in line) and ('end' not in line) and re.match(regex, line) and line:
                line = line.strip()
                if i>0:
                    print(i+59, line[1:], file=ff)
                i += 1
```
