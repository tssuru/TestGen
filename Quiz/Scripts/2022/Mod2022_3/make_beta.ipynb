```python
begin=True
with open('mod3.txt') as f, open('mod3.tex','w') as fo:
    for s in f:
        if s.startswith('%code'):
            if begin:
                s='\\begin{verbatim}\n'
            else:
                s='\\end{verbatim}\n'
            begin = not begin
        elif s.startswith('%answer'):
            s=''
        elif s.startswith('%beginitem'):
            s='\n\\newpage\n'
        else:
            s=s.replace('../','')
        
        fo.write(s)
            
```
